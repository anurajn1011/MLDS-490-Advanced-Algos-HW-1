# Value and Policy Iteration

```python
import os
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from src.q1.enumeration_methods import enumeration, value_iteration, policy_iteration
```

## Enumeration

```python
# Running with default params and mentioned assumptions
V_enumeration, policy_enumeration = enumeration()
```

```python
print("first dispatch state at t=0:", np.argmax(policy_enumeration[0]))
print("monotone V0:", np.all(np.diff(V_enumeration) >= 0))
```

```python
sns.set_theme(style="whitegrid", context="talk")

states = np.arange(len(V_enumeration))

fig, ax = plt.subplots(figsize=(9, 5))
sns.lineplot(x=states, y=V_enumeration, ax=ax)
ax.set(xlabel="Customers waiting",
       ylabel="$V_0(s)$",
       title="Optimal value function at t = 0 (Enumeration, T = 500)")
plt.tight_layout()
plt.show()
```

```python
V_enumeration[0]
```

## Value Iteration

```python
V_value, policy_value = value_iteration()
```

```python
print("first dispatch state at t=0:", np.argmax(policy_value))
print("monotone V0:", np.all(np.diff(V_value) >= 0))
```

```python
sns.set_theme(style="whitegrid", context="talk")

states = np.arange(len(V_value))

fig, ax = plt.subplots(figsize=(9, 5))
sns.lineplot(x=states, y=V_value, ax=ax)
ax.set(xlabel="Customers waiting",
       ylabel="$V_0(s)$",
       title="Optimal value function at t = 0 (Enumeration, T = Infinity)")
plt.tight_layout()
plt.show()
```
